```python
# there is only one cell with this name
adata = adata[adata.obs['annot_4'] != 'Unassigned Mixed (putative doublets)']
```

```python
plotting_utils.plot_umap(adata, 'annot_4', 10, 8, 10, True)
```

```python
temp = {
    'ambigious cbd':'marked_removal',
    'stromal cells cbd':'marked_removal',
    'Fibroblasts / stromal cells cbd':'marked_removal'
}
```

```python
adata.obs['marked_cells_1'] = adata.obs['annot_4'].map(temp)
adata.obs['marked_cells'] = adata.obs['marked_cells_1'].combine_first(adata.obs['marked_cells'])
```

```python
del adata.obs['marked_cells_1']
```

```python
plotting_utils.plot_umap(adata, 'annot_4', 10, 8, 10, True)
plotting_utils.plot_umap(adata, 'marked_cells', 10, 8, 10, True)
```
