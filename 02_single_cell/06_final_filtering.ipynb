# Final Filtering before annotation
* Added all the necessary information to the adata obs attribute.
* Filtering cells that have total counts of more than 95th percentile of the total counts data.
* Filtering cells that are marked as doublets by two or more doublet detection algorithms
    * Algorithms used are:
        * Doublet Detection
        * Scrublet
        * SOLO from scVI

```python
import os
# Project data root. Override with the ACC_DATA_ROOT environment variable.
ACC_DATA_ROOT = os.environ.get("ACC_DATA_ROOT", os.path.expanduser("~/acc_data"))
```

```python
import scanpy as sc
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
path = f'{ACC_DATA_ROOT}/outputs/h5ad_files/'
adata = sc.read_h5ad(path+'inprogress_2.h5ad')
adata
```

```python
upper_cutoff_95 = np.percentile(adata.obs["total_counts"], 95)
adata.obs["high_count_cell_95"] = adata.obs["total_counts"] > upper_cutoff_95
counter = 0
for i in list(adata.obs['high_count_cell_95']):
    if i == True:
        counter+=1
print('Number of cells with greater than 95 percentile count: ', counter)
```
